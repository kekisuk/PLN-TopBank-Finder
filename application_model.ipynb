```python
import pandas as pd
import spacy
from collections import Counter
import re
```

Leitura do Arquivo e Pré-processamento

```python
# Carregar o modelo do spaCy para NLP
nlp = spacy.load("pt_core_news_sm")

# Leitura do arquivo CSV ou JSON com os posts
df_banks = pd.read_csv('database/posts_banks_structured.csv')
```

```python
# Exemplo de pré-processamento de texto
def preprocess_text(text):
    # Remover caracteres especiais e números
    text = re.sub(r'[^a-zA-Záéíóúàâêîôûãõç\s]', '', text)
    
    # Tokenizar o texto e aplicar lematização
    doc = nlp(text.lower())  # Transformar o texto em minúsculas
    tokens = [token.lemma_ for token in doc if not token.is_stop]  # Remover stopwords e lematizar
    return tokens
```

```python
# Pré-processar os textos dos posts
df_banks['tokens'] = df_banks['pergunta'].apply(preprocess_text)
```
